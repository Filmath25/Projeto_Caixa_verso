Extração e Tratamento de Dados de Rendimento - IBGE (Estados).

IMPORTACOES

```python
import pandas as pd
import sidrapy
```

Dicionário de Variáveis
- **5929:** Rendimento médio mensal nominal das pessoas de 14 anos ou mais
- **5933:** Rendimento médio mensal real das pessoas de 14 anos ou mais
- **5941:** Variação relativa do rendimento médio real (comparação com período anterior)

REQUISICAO

```python
# https://apisidra.ibge.gov.br/values/t/4660/n3/all/v/all/p/all/d/v5937%201,v5941%201

data = sidrapy.get_table(table_code="4660", territorial_level="3", ibge_territorial_code="all", period="all", variable='all')
```

INFO DOS DADOS

```python
data.info()
```

TRANSFORMACAO

```python
data.columns = data.iloc[0]
```

```python
data.drop(0, inplace=True)
```

```python
data = data.drop(columns=['Nível Territorial (Código)', 'Nível Territorial', 'Ano', 'Unidade de Medida (Código)', 'Unidade da Federação (Código)','Unidade de Medida'])
```

```python
data['Valor'] = data['Valor'].apply(lambda x: pd.to_numeric(x, errors='coerce'))
```

```python
print(f'valores convertidos para nan: {data['Valor'].isna().sum()}')
```

```python
data = data.dropna(subset='Valor')
```

```python
data.info()
```

IMPORTANDO PARA CSV COM ENCODING

```python
csv = data.to_csv('rendimento_mensal_estados_brasil.csv', index=False, encoding='utf-8-sig')
print(f'O arquivo csv foi criado')
```
